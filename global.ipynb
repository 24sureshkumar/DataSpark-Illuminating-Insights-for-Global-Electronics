# (LEFT) Merge dataset on sales and exchange To Calculate Total revenue USD -------- 

```python
chunk_size = 3000  # Adjust this number based on your available memory
sales=pd.read_csv(r"D:\mainboot\gendral\data\sales_data.csv", chunksize=chunk_size)

for chunk in sales:
    # Process each chunk as needed
    merged_chunk = pd.merge(chunk, exchange, on='Currency Code', how='outer')
    merged_chunk = pd.merge(merged_chunk, product[['ProductKey', 'Unit Price USD']], on='ProductKey', how='left')
    merged_chunk['Total Revenue USD'] = merged_chunk['Quantity'] * merged_chunk['Unit Price USD'] / merged_chunk['Exchange']
```

```python
sales["Date"]=pd.to_datetime(sales["Date"],errors="coerce")
sales["Order Date"]=pd.to_datetime(sales["Order Date"],errors="coerce")
```

```python
# Average order value----------
sales["Order value"]=sales["Quantity"]*sales["Unit Price USD"]
average_order_value = sales.groupby('CustomerKey')['Order value'].mean()
print("\nAverage Order Value per Customer:")
print(average_order_value.describe())
se.histplot(average_order_value, bins=20, kde=True)
py.title("Average Order Value Distribution")
py.ylabel('Frequency of customers')
py.show()
```

```python
# Freqency of purchase-----
frequency_purchase=sales.groupby("CustomerKey")["Order Number"].nunique()
print(frequency_purchase.reset_index())
py.hist(frequency_purchase,bins=20,color="lightgreen",edgecolor='black')
py.xlabel('Number of unique orders')
py.ylabel('Number of customers')
py.show()

```

```python
# Most preferred product------
preferred_product=sales["ProductKey"].value_counts().head(10)
print(preferred_product)
py.figure(figsize=(10, 6))
py.bar(preferred_product.index.astype(str), preferred_product.values, color='lightcoral')  # Convert index to string for better readability
py.title('Top 10 Preferred Products', fontsize=14)
py.xlabel('Product Key', fontsize=12)
py.ylabel('Number of Orders', fontsize=12)
py.xticks(rotation=45)  # Rotate x-axis labels for better readability
py.show()
```

```python
merged_data=pd.merge(customer[["CustomerKey","age"]],sales[["CustomerKey","Order value","Order Number"]],on="CustomerKey")
merged_data.info()
merged_data
```

```python
# Assuming 'sales_data' contains customer demographic data like Age, Gender, etc.
# Segment customers by age group and purchasing behavior

# Example segmentation by age
bins = [18, 30, 45, 60, 100]
labels = ['18-29', '30-44', '45-59', '60+']
merged_data['Age Group'] = pd.cut(merged_data['age'], bins=bins, labels=labels)

# Aggregate by Age Group
sales_by_age_group = merged_data.groupby('Age Group').agg({
    'Order value': 'sum',
    'Order Number': 'nunique'
}).reset_index()
print(sales_by_age_group)
# Plotting sales by age group
py.figure(figsize=(10, 5))
py.bar(sales_by_age_group['Age Group'], sales_by_age_group['Order value'], color='lightblue')
py.title('Sales by Age Group')
py.xlabel('Age Group')
py.ylabel('Total Sales')
py.show()
```

```python
merged_data1=pd.merge(sales[['ProductKey','Quantity']],product[['ProductKey','Product Name','Category','Subcategory','Unit Cost USD','Unit Price USD']],on='ProductKey')
merged_data1
```

```python
# product popularity-----
product_popularity=merged_data1.groupby(['ProductKey','Product Name'])["Quantity"].sum().reset_index()
print(product_popularity)
most_popularity_product=product_popularity.sort_values(by='Quantity',ascending=False).head(10)
print('most popularity product: \n',most_popularity_product)
latest_popularity_product=product_popularity.sort_values(by='Quantity',ascending=True).head(10)
print('latest popularity product : \n',latest_popularity_product)
se.barplot(x='Quantity',y='Product Name',data=most_popularity_product,palette='Blues_r')
py.title("Top 10 Most popularity product")
```

```python
se.barplot(x='Quantity',y='Product Name',data=latest_popularity_product,palette='Reds_r')
py.title("Top 10 Latest popularity product")
py.show()
```

```python
# profit margine---------
merged_data1['profit']=merged_data1['Unit Price USD'] - merged_data1['Unit Cost USD'] * merged_data1['Quantity']
# calculate total margine------
profitability=merged_data1.groupby(["ProductKey",'Product Name']).agg({'profit': 'sum' , 'Unit Price USD' : 'mean','Unit Cost USD' : 'mean'}).reset_index()
profitability['total margine']=(profitability['Unit Price USD'] - profitability['Unit Cost USD']) / (profitability['Unit Price USD']) * 100
top10=profitability.sort_values(by='profit',ascending=False).head(10)
py.figure(figure=((16,12)))
se.barplot(x='profit',y='Product Name',hue='ProductKey',data=top10,color='lightblue')
py.title(' Top 10 Most profitable product')
py.xlabel('Total Profit (USD)', fontsize=14)
py.ylabel('Product Name', fontsize=14)
py.yticks(rotation=0)
py.tight_layout()
py.show()
```

```python
print(top10)
se.barplot(x='total margine',y='Product Name',hue='ProductKey',data=top10,palette='coolwarm')
py.title('Top 10 product by Total margine')
py.show()
```

```python
# category analysis--------
category_analysis=merged_data1.groupby(['Category','Subcategory']).agg({'Quantity':'sum'}).reset_index()
top_products=category_analysis.sort_values(by='Quantity',ascending=False).head(10)
print(top_products)
se.barplot(x='Quantity',y='Subcategory',hue='Category',data=top_products,palette='coolwarm')
py.title('Top 10 product category')
py.show()
```
