```python
import mysql.connector
my=mysql.connector.connect(
    host='localhost',
    user='root',
    password='',
    database='Global_electronic'
    )
mycursor=my.cursor(buffered=True)
```

```python
import pandas as pd
product=pd.read_csv(r"D:\mainboot\gendral\data\product_data.csv")
sales=pd.read_csv(r"D:\mainboot\gendral\data\sales_data.csv")
customer=pd.read_csv(r"D:\mainboot\gendral\data\new customer data.csv")
store=pd.read_csv(r"D:\mainboot\gendral\data\stores_data.csv")
exchange=pd.read_csv(r"D:\mainboot\gendral\data\Exchange_Rates_data.csv")
```

```python
mycursor.execute('show databases')
for i in mycursor:
    print(i)
```

```python
mycursor.execute('create database Global_electronic')
my.commit()
```

```python
# Table creation ---------
mycursor.execute('''create table Global_electronic.product(
                 ProductKey int(6),
                 Product_Name text(100),
                 Brand varchar(15),
                 Color varchar(10),
                 Unit_Cost_USD decimal(5,2),
                 Unit_Price_USD decimal(5,2),
                 SubcategoryKey int(8),
                 Subcategory varchar(70),
                 CategoryKey int(10),       
                 Category varchar(50)
                  )''')
```

```python
# Insert query -------------
product_data=(''' insert into Global_electronic.product( 
              ProductKey,
              Product_Name,
              Brand,
              Color,
              Unit_Cost_USD,
              Unit_Price_USD,
              SubcategoryKey,
              Subcategory,
              CategoryKey,
              Category) values (%s,%s,%s,%s,%s,%s,%s,%s,%s,%s)''')
data1=product.values.tolist()
mycursor.executemany(product_data,data1)
my.commit()
```

```python
mycursor.execute('''create table Global_electronic.sales(
                 Order_Number int,
                 Line_Item int,
                 Order_Date date,
                 CustomerKey int,
                 StoreKey int,
                 ProductKey int,
                 Quantity int,
                 Currency_Code varchar(10)              
                 )''')
```

```python
sales_data=('''insert into Global_electronic.sales(
                Order_Number,
                Line_Item,
                Order_Date,
                CustomerKey,
                StoreKey,
                ProductKey,
                Quantity,
                Currency_Code) values (%s,%s,%s,%s,%s,%s,%s,%s)''')


for _, row in sales.iterrows():
    mycursor.execute(sales_data, tuple(row))

# Commit the transaction
my.commit()
```

```python
mycursor.execute('''create table Global_electronic.customer (
               CustomerKey int,
               Gender varchar(225),
               Name varchar(225),
               City varchar(225),
               `State Code` varchar(225),
               State varchar(225),
               `Zip Code` text,    
               Country varchar(225),  
               Continent varchar(225),     
               Birthday date,
               age int)''')
```

```python
customer_data=('''insert into Global_electronic.customer(
               CustomerKey,
               Gender,
               Name,
               City,
               `State Code`,
               State,
               `Zip Code`,
               Country,
               Continent,
               Birthday,
               age) values (%s,%s,%s,%s,%s,%s,%s,%s,%s,%s,%s)''')
for _, row in customer.iterrows():
    mycursor.execute(customer_data, tuple(row))
# Commit the transaction
my.commit()
```

```python
mycursor.execute('''create  table Global_electronic.store(
                 StoreKey int,
                 Country varchar(225),
                 State varchar(225),
                 `Square Meters` decimal(5,2),
                 `Open Date` date)''')
```

```python
store_data=('''insert into Global_electronic.store(
            StoreKey,
            Country,
            State,
            `Square Meters`,
            `Open Date`) values (%s,%s,%s,%s,%s)''')
for i,row in store.iterrows():
    mycursor.execute(store_data,tuple(row))
my.commit()
```

```python
mycursor.execute('''create table Global_electronic.exchange (
                 Date date,
                 `Currency Code` varchar(225),
                 Exchange float(8,2))''')
```

```python
exchange_data=('''insert into Global_electronic.exchange(
               Date,
               `Currency Code`,
               Exchange) values (%s,%s,%s)''')
for j,k in exchange.iterrows():
    mycursor.execute(exchange_data,tuple(k))
my.commit()
```
